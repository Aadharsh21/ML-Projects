```python
!pip install opendatasets --quiet
import pandas as pd
import os
import numpy as np
import opendatasets as od
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, InputLayer, BatchNormalization
from tensorflow.keras.callbacks import ModelCheckpoint
from sklearn.impute import SimpleImputer
from
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import classification_report
```

```python
dataset_url = 'https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud?select=creditcard.csv'
```

```python
od.download(dataset_url)
```

```python
data_dir = './creditcardfraud'
os.listdir(data_dir)
train_csv = data_dir + '/creditcard.csv'
```

```python
carddf = pd.read_csv(train_csv)
```

```python
carddf
```

```python
carddf.info()
```

```python
newdf = carddf.copy()
```

```python
newdf['Amount'] = RobustScaler().fit_transform(newdf['Amount'].values.reshape(-1,1))
```

```python
newdf.describe()
```

```python
newdf['Time'] = (newdf['Time'] - newdf['Time'].min()) / (newdf['Time'].max() - newdf['Time'].min())
```

```python
newdf = newdf.sample(frac=1, random_state = 1)
```

```python
train_valdf, testdf = train_test_split(newdf, test_size = 0.1, random_state =42)
traindf, valdf = train_test_split(train_valdf,test_size = 0.1111111, random_state = 42 )
```

```python
testdf.shape, valdf.shape, traindf.shape
```

```python
input_cols = list(traindf.columns)[:-1]
target_col = 'Class'
```

```python
traininputs = traindf[input_cols].copy()
traintargets = traindf[target_col].copy()
valinputs = valdf[input_cols].copy()
valtargets = valdf[target_col].copy()
testinputs = testdf[input_cols].copy()
testtargets = testdf[target_col].copy()
valtargets = tf.cast(valtargets, tf.int64)
```

```python
the_nn = Sequential()
the_nn.add(InputLayer((traininputs.shape[1],)))
the_nn.add(Dense(2, activation = 'relu'))
the_nn.add(BatchNormalization())
the_nn.add(Dense(1, activation = 'sigmoid'))

checkpoint = ModelCheckpoint('the_nn.keras',save_best_only=True)
the_nn.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
```

```python
the_nn.summary()
```

```python
the_nn.fit(traininputs, traintargets, validation_data = (valinputs, valtargets), epochs = 5, callbacks = [checkpoint])
```

```python
print(classification_report(valtargets, (the_nn.predict(valinputs) ).astype(int)))
```
